```python
def balik_kata(teks):
    stack = []
    
    # Memasukkan setiap huruf ke dalam stack (Push)
    for huruf in teks:
        stack.append(huruf)
        
    hasil = ""
    # Mengambil huruf dari stack satu per satu (Pop)
    while stack:
        hasil += stack.pop()
        
    return hasil

while True:
    print("\n=== PROGRAM BALIK KATA ===")
    print("1. Balik kata")
    print("2. Keluar")
    
    pilihan = input("Pilih menu: ")
    
    if pilihan == "1":
        teks = input("Masukkan kata/kalimat: ")
        hasil = balik_kata(teks)
        print("Hasil:", hasil)
        
    elif pilihan == "2":
        print("Program selesai")
        break
        
    else:
        print("Pilihan tidak valid")
```

```python
#Celebrity Problem
def knows(a, b, mat):
    return mat[a][b] == 1

def find_celebrity(mat):
    n = len(mat)
    stack = list(range(n))

    # STEP 1: Eliminasi kandidat
    while len(stack) > 1:
        a = stack.pop()
        b = stack.pop()

        if knows(a, b, mat):
            # Jika a kenal b, maka a bukan selebriti, b mungkin selebriti
            stack.append(b)
        else:
            # Jika a tidak kenal b, maka b bukan selebriti, a mungkin selebriti
            stack.append(a)

    # kandidat tersisa
    candidate = stack.pop()

    # STEP 2: Verifikasi
    for i in range(n):
        if i != candidate:
            # Selebriti tidak boleh kenal siapapun (knows(candidate, i))
            # Dan semua orang harus kenal selebriti (not knows(i, candidate))
            if knows(candidate, i, mat) or not knows(i, candidate, mat):
                return -1

    return candidate

# Matriks pertemanan (Contoh)
mat = [
    [0, 1, 1],
    [0, 0, 1],
    [0, 0, 0]
]

print(find_celebrity(mat))
```

```python
rod = ['S', 'A', 'D']
stacks = [[], [], []]

def moveDisk(a, b):

    if not stacks[b] or (stacks[a] and stacks[a][-1] < stacks[b][-1]):
        print(f"Move disk {stacks[a][-1]} from rod {rod[a]} to rod {rod[b]}")
        stacks[b].append(stacks[a].pop())
    else:
        
        moveDisk(b, a)

def towerOfHanoi(n):
    print(f"Tower of Hanoi for {n} disks:")
    
    src, aux, dest = 0, 1, 2
    
    stacks[src] = list(range(n, 0, -1))
    
    totalMoves = (1 << n) - 1 
    
    if n % 2 == 0:
        aux, dest = dest, aux
        
    for i in range(1, totalMoves + 1):
        if i % 3 == 0:
            moveDisk(aux, dest)
        elif i % 3 == 1:
            moveDisk(src, dest)
        else:
            moveDisk(src, aux)

if __name__ == "__main__":
    n = 3  # Jumlah disk
    towerOfHanoi(n)
```

```python
class myQueue:
    def __init__(self):
        self.s1 = []
        self.s2 = []

    # Enqueue operation (costly)
    def enqueue(self, x):
        
        # Move all elements from s1 to s2
        while self.s1:
            self.s2.append(self.s1.pop())
            
        # Push the new item into s1
        self.s1.append(x)
        
        # Move everything back to s1
        while self.s2:
            self.s1.append(self.s2.pop())

    # Dequeue operation
    def dequeue(self):
        if not self.s1:
            # Queue underflow
            return
        return self.s1.pop()
    
    # Front operation
    def front(self):
        if not self.s1:
            return -1
        return self.s1[-1]

    # Size operation
    def size(self):
        return len(self.s1)

if __name__ == "__main__":
    q = myQueue()
    q.enqueue(1)
    q.enqueue(2)
    q.enqueue(3)

    print("Front:", q.front())
    print("Size:", q.size())

    q.dequeue()
    print("Front:", q.front())
    print("Size:", q.size())
```
