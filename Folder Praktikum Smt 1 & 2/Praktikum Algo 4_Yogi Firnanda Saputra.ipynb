# *Linked List*
*Linked List* (Daftar Berantai) adalah struktur data linear di mana elemen-elemen (disebut node) saling terhubung menggunakan pointer / referensi. Berbeda dengan array, data pada linked list tidak disimpan bersebelahan di memori.  

Nah, berikut merupakan Contoh Code Pemrograman Berbasis Python yang menggunakan *Linked List*

```python
# latihan
class Queue:
    def __init__(self):
        self.queue = {}

    def enqueue(self, nomor, nama):
        self.queue[nomor] = nama
        print("Nomor antrian", nomor, "ditambahkan untuk", nama)

    def dequeue(self):
        if len(self.queue) == 0:
            print("Antrian kosong")
        else:
            nomor = list(self.queue.keys())[0]
            nama = self.queue[nomor]
            del self.queue[nomor]
            print("Nomor antrian", nomor, "-", nama, "dipanggil")

    def tampilkan(self):
        if len(self.queue) == 0:
            print("Tidak ada antrian")
        else:
            print("Daftar Antrian:")
            for nomor, nama in self.queue.items():
                print("Nomor:", nomor, "| Nama:", nama)


antrian = Queue() 

while True:
    print("\n=== SISTEM ANTRIAN MAHASISWA ===")
    print("1. Tambah Antrian")
    print("2. Panggil Antrian")
    print("3. Tampilkan Antrian")
    print("4. Keluar")

    pilihan = int(input("Pilih menu: "))

    if pilihan == 1:
        nomor = input("Masukkan nomor antrian: ")
        nama = input("Masukkan nama mahasiswa: ")
        antrian.enqueue(nomor, nama)

    elif pilihan == 2:
        antrian.dequeue()

    elif pilihan == 3:
        antrian.tampilkan()

    elif pilihan == 4:
        print("Program selesai")
        break

    else:
        print("Pilihan tidak tersedia")
```

# *Analisis Notasi Big O*
Pada pemrograman tersebut saya menganalisis bahwa notasi Big O pada pemrograman tersebut ada yang menggunakan Notasi Linear dan Konstan yang saya dapat. Diantara lain :  

1. enqueue()
2. dequeue()
3. tampilkan()  

Berikut merupakan penjelasan pada notasi tersebut :  

# 1. *Operasi enqueue()*

*self.queue[nomor] = nama*  

Operasi ini hanya:  
menambahkan satu elemen ke dictionary, tidak bergantung pada jumlah data sebelumnya

Kompleksitas waktunya:  
O(1) — Konstan  
*Artinya waktu eksekusi tetap sama walaupun jumlah antrian bertambah banyak.*

# 2. *Operasi dequeue()*

*nomor = list(self.queue.keys())[0]*

Di sini terjadi dua proses:  
1. self.queue.keys() → mengambil semua key  
2. list(...) → mengubahnya menjadi list

Jika ada n data, maka Python harus membaca semua key.

Kompleksitas:  
O(n) — Linear

Kemudian:  
*del self.queue[nomor]*

Operasi delete pada dictionary adalah:  
O(1)

Tetapi karena ada proses list(self.queue.keys()), maka totalnya tetap:  
O(n)

# 3. *Operasi tampilkan()*

*for nomor, nama in self.queue.items():*

Loop ini membaca semua data dalam antrian. Jika ada n mahasiswa, maka loop berjalan n kali.

Kompleksitas:  
O(n) — Linear

# 4. *Struktur Menu while True*
Loop ini tidak mempengaruhi kompleksitas data, karena hanya menjalankan menu sampai user keluar.

# *Kesimpulan*
Program ini menggunakan kombinasi notasi Big O, yaitu:  
O(1) — Konstan pada operasi enqueue  
O(n) — Linear pada operasi dequeue dan tampilkan

Namun jika dilihat dari operasi yang paling berat, maka kompleksitas program cenderung linear O(n).

| Operasi   | Kompleksitas       |
| --------- | ------------------ |
| enqueue   | **O(1)** (Konstan) |
| dequeue   | **O(n)** (Linear)  |
| tampilkan | **O(n)** (Linear)  |

Program ini tidak murni konstan, karena terdapat operasi linear O(n) pada dequeue() dan tampilkan().
