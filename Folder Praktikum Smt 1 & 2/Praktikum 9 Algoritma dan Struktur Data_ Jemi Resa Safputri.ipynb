```python
# ==========================================
# INISIALISASI DATA (Minimal 5 Mahasiswa)
# ==========================================
data_mahasiswa = [
    {"nama": "Jemi Resa Safputri", "nim": "2511102441174", "nilai": 98},
    {"nama": "Yogi Firnanda Saputra", "nim": "2511102441170", "nilai": 92},
    {"nama": "Wafiq Azizah", "nim": "2511102441182", "nilai": 68},
    {"nama": "Prabu sanjaya", "nim": "2511102443141", "nilai": 75},
    {"nama": "Abu hamka Sucahya", "nim": "2511102446263", "nilai": 88}
]

# ==========================================
# FUNGSI SORTING (BUBBLE SORT)
# ==========================================
def urutkan_nilai_tertinggi(arr):
    n = len(arr)
    # Lakukan perulangan sebanyak jumlah data
    for i in range(n):
        swapped = False
        # Perulangan untuk membandingkan elemen yang bersebelahan
        for j in range(0, n - i - 1):
            # Jika nilai elemen ke-j LEBIH KECIL dari elemen ke-(j+1), tukar posisinya
            if arr[j]["nilai"] < arr[j + 1]["nilai"]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        
        # Jika tidak ada elemen yang ditukar, berarti data sudah urut
        if not swapped:
            break
            
    return arr

# ==========================================
# OUTPUT PROGRAM
# ==========================================
print("=== DATA MAHASISWA SEBELUM DIURUTKAN ===")
for mhs in data_mahasiswa:
    print(f"NIM: {mhs['nim']} | Nama: {mhs['nama']:<15} | Nilai: {mhs['nilai']}")

# Memanggil fungsi sorting
data_terurut = urutkan_nilai_tertinggi(data_mahasiswa)

print("\n=== DATA MAHASISWA SETELAH DIURUTKAN (NILAI TERTINGGI) ===")
for mhs in data_terurut:
    print(f"NIM: {mhs['nim']} | Nama: {mhs['nama']:<15} | Nilai: {mhs['nilai']}")
```

# Analisis Program Pengurutan Data Mahasiswa  
Struktur analisis ini disusun secara sistematis agar mudah dipahami dari sisi struktur data maupun efisiensi algoritma.

## 1. Tujuan dan Fungsi Program
Program ini dibuat untuk mengurutkan sekumpulan data mahasiswa berdasarkan atribut nilai secara Descending (dari nilai tertinggi ke terendah). Data asal yang tidak beraturan diproses sehingga mahasiswa dengan pencapaian tertinggi berada di urutan pertama.

## 2. Analisis Struktur Data
Data mahasiswa disimpan menggunakan tipe data List of Dictionaries di Python.  

* List (Array): Digunakan sebagai wadah utama (data_mahasiswa) yang menampung seluruh elemen.  

* Dictionary: Setiap entitas mahasiswa direpresentasikan dengan struktur key-value (nama, nim, nilai). Pendekatan ini sangat efektif untuk mengelompokkan berbagai atribut tipe data yang berbeda ke dalam satu objek tunggal yang mudah diakses.

## 3. Analisis Logika Algoritma (Bubble Sort)
Algoritma yang digunakan adalah iterasi Bubble Sort yang telah dioptimasi. Berikut adalah cara kerjanya:

* Perulangan Luar (for i in range(n)): Menentukan jumlah maksimal iterasi (pass) yang dibutuhkan, yaitu sebanyak jumlah elemen array.  

* Perulangan Dalam (for j in range(0, n - i - 1)): Melakukan perbandingan terhadap elemen yang posisinya bersebelahan. Batas atas dikurangi i karena pada setiap lintasan, elemen terkecil pada iterasi tersebut sudah pasti berada di posisi paling akhir yang benar, sehingga tidak perlu dievaluasi ulang.  

* Logika Penukaran (Swap): Kondisi if arr[j]["nilai"] < arr[j + 1]["nilai"] mengecek apakah nilai saat ini lebih kecil dari nilai berikutnya. Jika benar, kedua posisi elemen ditukar. Ini memastikan nilai yang lebih kecil "tenggelam" ke belakang, dan nilai yang lebih besar "mengapung" ke depan.  

* Optimasi (swapped): Variabel boolean swapped bertindak sebagai penanda. Jika dalam satu lintasan perulangan dalam tidak ada satu pun elemen yang ditukar (if not swapped: break), program akan langsung menghentikan proses iterasi. Hal ini menandakan bahwa seluruh data sudah dalam keadaan terurut.  

## 4. Kompleksitas Algoritma (Big-O Notation)
Kompleksitas Waktu (Time Complexity):

* Best Case: O(n) — Terjadi apabila sekumpulan data tersebut sudah dalam keadaan terurut menurun sejak awal. Program hanya melakukan perbandingan satu lintasan penuh tanpa ada pertukaran, lalu langsung memicu break.  

* Worst Case & Average Case: O(n2) — Terjadi jika sekumpulan data acak atau terurut terbalik (dari kecil ke besar). Algoritma harus menjalankan perulangan bersarang secara penuh karena pertukaran terus terjadi di setiap lintasan.  

* Kompleksitas Ruang (Space Complexity): O(1) — Algoritma ini memproses elemen langsung di dalam tempatnya (in-place sorting). Tidak ada alokasi memori dinamis tambahan yang besarnya bergantung pada jumlah data n. Pertukaran nilai hanya membutuhkan sedikit memori sementara dari sistem.
