```python
import time
import random

# --- 1. Bubble Sort ---
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                # Tukar posisi jika elemen saat ini lebih besar dari selanjutnya
                arr[j], arr[j+1] = arr[j+1], arr[j]

# --- 2. Selection Sort ---
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[min_idx] > arr[j]:
                min_idx = j
        # Tukar elemen terkecil yang ditemukan dengan elemen pertama yang belum terurut
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

# --- 3. Insertion Sort ---
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i-1
        # Pindahkan elemen yang lebih besar dari key ke satu posisi di depannya
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

# --- Pengujian dan Pengukuran Waktu ---
if __name__ == "__main__":
    # Membuat array acak berisi 1000 elemen
    data_asli = [random.randint(1, 10000) for _ in range(1000)]
    
    # Copy data agar setiap algoritma mengurutkan data awal yang sama
    data_bubble = data_asli.copy()
    data_selection = data_asli.copy()
    data_insertion = data_asli.copy()

    print("Mengurutkan 1000 data acak...\n")

    # Uji Bubble Sort
    start_time = time.time()
    bubble_sort(data_bubble)
    print(f"Waktu Bubble Sort    : {time.time() - start_time:.6f} detik")

    # Uji Selection Sort
    start_time = time.time()
    selection_sort(data_selection)
    print(f"Waktu Selection Sort : {time.time() - start_time:.6f} detik")

    # Uji Insertion Sort
    start_time = time.time()
    insertion_sort(data_insertion)
    print(f"Waktu Insertion Sort : {time.time() - start_time:.6f} detik")
```

```python
def coin_change_greedy(coins, amount):
    # Urutkan koin dari nominal paling besar ke kecil
    coins.sort(reverse=True)
    hasil = []
    
    for koin in coins:
        # Selama sisa uang masih bisa dikurangi dengan koin saat ini
        while amount >= koin:
            amount -= koin
            hasil.append(koin)
            
    # Jika masih ada sisa amount yang tidak bisa dikembalikan koin
    if amount > 0:
        return "Tidak ada solusi pas untuk koin yang tersedia."
    
    return hasil

# --- Pengujian ---
koin_tersedia = [1000, 500, 200, 100]
uang_kembalian = 2700

hasil_greedy = coin_change_greedy(koin_tersedia, uang_kembalian)
print(f"Kembalian untuk {uang_kembalian}:")
print(f"Koin yang digunakan (Greedy) : {hasil_greedy}")
print(f"Total koin (Greedy)          : {len(hasil_greedy) if isinstance(hasil_greedy, list) else '-'}")
```

```python
def coin_change_exhaustive(coins, amount):
    # Fungsi rekursif untuk mencoba semua kombinasi
    def cari_kombinasi(sisa_uang, kombinasi_saat_ini):
        # Basis kasus: jika uang pas 0, kembalikan kombinasi
        if sisa_uang == 0:
            return kombinasi_saat_ini
        # Jika sisa uang negatif, maka jalur kombinasi ini salah
        if sisa_uang < 0:
            return None
            
        kombinasi_terbaik = None
        
        # Coba tambahkan setiap jenis koin satu per satu
        for koin in coins:
            hasil_rekursif = cari_kombinasi(sisa_uang - koin, kombinasi_saat_ini + [koin])
            
            # Jika ditemukan solusi di jalur ini
            if hasil_rekursif is not None:
                # Update jika ini adalah solusi pertama, atau jika jumlah koinnya lebih sedikit dari yang terbaik sebelumnya
                if kombinasi_terbaik is None or len(hasil_rekursif) < len(kombinasi_terbaik):
                    kombinasi_terbaik = hasil_rekursif
                    
        return kombinasi_terbaik

    hasil_akhir = cari_kombinasi(amount, [])
    if hasil_akhir is None:
         return "Tidak ada solusi pas."
    return hasil_akhir

# --- Pengujian ---
# Catatan: Exhaustive search akan sangat lambat jika angkanya besar. 
# Kita gunakan angka kecil untuk demonstrasi.
koin_tersedia_ex = [1, 3, 4]
uang_kembalian_ex = 6

print(f"\nSistem Koin: {koin_tersedia_ex}, Ingin kembalian: {uang_kembalian_ex}")

# Menguji Exhaustive Search
hasil_ex = coin_change_exhaustive(koin_tersedia_ex, uang_kembalian_ex)
print(f"Koin yang digunakan (Exhaustive) : {hasil_ex}")
print(f"Total koin (Exhaustive)          : {len(hasil_ex)}")

# Sebagai perbandingan, mari kita lihat hasil Greedy di kasus yang sama
hasil_gr = coin_change_greedy(koin_tersedia_ex.copy(), uang_kembalian_ex)
print(f"Koin yang digunakan (Greedy)     : {hasil_gr}")
print(f"Total koin (Greedy)              : {len(hasil_gr)}")
```
