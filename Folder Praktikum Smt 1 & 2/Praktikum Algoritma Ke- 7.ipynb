```python
import matplotlib.pyplot as plt

# 1. DEFINISI STRUKTUR DATA
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# 2. FUNGSI OPERASI BST (INSERT & SEARCH)
def insert(root, data):
    if root is None:
        return Node(data)
    if data < root.data:
        root.left = insert(root.left, data)
    else:
        root.right = insert(root.right, data)
    return root

def search(root, key):
    if root is None:
        return False
    if root.data == key:
        return True
    if key < root.data:
        return search(root.left, key)
    else:
        return search(root.right, key)

# 3. FUNGSI ANALISIS PROPERTI TREE
def height(root):
    if root is None:
        return -1
    return max(height(root.left), height(root.right)) + 1

def count_nodes(root):
    if root is None:
        return 0
    return 1 + count_nodes(root.left) + count_nodes(root.right)

def count_leaf(root):
    if root is None:
        return 0
    if root.left is None and root.right is None:
        return 1
    return count_leaf(root.left) + count_leaf(root.right)

# 4. FUNGSI TRAVERSAL
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)

def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)

def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")

# 5. FUNGSI VISUALISASI
def plot_tree(node, x, y, dx):
    if node is None:
        return

    # Gambar lingkaran node
    plt.text(x, y, str(node.data), ha='center', va='center',
             bbox=dict(boxstyle="circle", ec="black", fc="lightblue"))

    # Garis ke anak kiri
    if node.left:
        plt.plot([x, x - dx], [y, y - 1], color="black", zorder=1)
        plot_tree(node.left, x - dx, y - 1, dx / 2)

    # Garis ke anak kanan
    if node.right:
        plt.plot([x, x + dx], [y, y - 1], color="black", zorder=1)
        plot_tree(node.right, x + dx, y - 1, dx / 2)

# 6. MAIN PROGRAM
if __name__ == "__main__":
    root = None
    data_list = [10, 5, 15, 3, 7, 12, 18]

    # Memasukkan data ke dalam BST
    for data in data_list:
        root = insert(root, data)

    # Tampilkan Informasi Statistik
    print("--- Statistik Tree ---")
    print("Jumlah node :", count_nodes(root))
    print("Jumlah leaf :", count_leaf(root))
    print("Tinggi tree :", height(root))
    
    key = 7
    print(f"Cari {key}      :", "Ditemukan" if search(root, key) else "Tidak ditemukan")

    # Tampilkan Hasil Traversal
    print("\n--- Hasil Traversal ---")
    print("Preorder  :", end=" ")
    preorder(root)
    print("\nInorder   :", end=" ")
    inorder(root)
    print("\nPostorder :", end=" ")
    postorder(root)
    print("\n")

    # Visualisasi dengan Matplotlib
    plt.figure(figsize=(8, 6))
    plot_tree(root, 0, 0, 2)
    plt.axis('off')
    plt.title("Visualisasi Binary Search Tree")
    plt.show()
```

# 1. Mekanisme Penyimpanan (BST Property)  
Berbeda dengan pohon biner biasa, kode ini menggunakan fungsi insert yang menerapkan aturan Binary Search Tree. Setiap data baru dibandingkan dengan root: jika lebih kecil akan masuk ke cabang kiri, dan jika lebih besar akan masuk ke cabang kanan. Hal ini membuat pencarian data (search) menjadi sangat efisien karena algoritma tidak perlu memeriksa seluruh node, melainkan cukup mengikuti jalur yang sesuai dengan nilai kunci.  

# 2. Analisis Struktural dan Kuantitatif  
Kode ini memiliki kemampuan untuk membedah anatomi pohon melalui tiga fungsi rekursif:  
- count_nodes: Menghitung seluruh elemen yang ada di dalam pohon.

- count_leaf: Mengidentifikasi "daun" atau node terminal yang tidak memiliki anak, yang penting untuk memahami lebar dasar pohon.

- height: Menentukan kedalaman atau level maksimum pohon, yang secara langsung memengaruhi performa pencarian.  

# 3. Penelusuran Hierarki (Traversal)  
Dengan menyertakan tiga metode traversal, kode ini menunjukkan cara berbeda dalam mengekstraksi data:  

- Preorder: Berguna untuk menduplikasi struktur pohon.

- Inorder: Karena ini adalah BST, hasil traversal ini akan selalu menghasilkan urutan angka yang terurut menaik (sorted), yang sangat berguna untuk validasi data.

- Postorder: Sering digunakan untuk proses penghapusan pohon atau evaluasi ekspresi matematika.  

# 4. Visualisasi Grafis  
Salah satu keunggulan utama kode ini adalah penggunaan Matplotlib. Fungsi plot_tree mengubah abstraksi kode menjadi bentuk visual. Penggunaan parameter dx yang berkurang secara geometris ($dx/2$) memastikan bahwa meskipun pohon memiliki banyak cabang, tata letak visualnya tetap proporsional dan menjaga estetika presentasi data.
