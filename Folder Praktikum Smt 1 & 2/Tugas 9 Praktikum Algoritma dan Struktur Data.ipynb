```python
# ==========================================
# INISIALISASI DATA (Minimal 5 Mahasiswa)
# ==========================================
data_mahasiswa = [
    {"nama": "Yogi Firnanda Saputra", "nim": "2511102441170", "nilai": 85},
    {"nama": "Jemi Resa Safputri", "nim": "2511102441174", "nilai": 92},
    {"nama": "Budi Utomo Sucahyo", "nim": "2511102441182", "nilai": 78},
    {"nama": "Hanif Anugerah Ramadhan", "nim": "2511102441270", "nilai": 95},
    {"nama": "Joko Widodo", "nim": "2511102441413", "nilai": 88}
]

# ==========================================
# FUNGSI SORTING (BUBBLE SORT)
# ==========================================
def urutkan_nilai_tertinggi(arr):
    n = len(arr)
    # Lakukan perulangan sebanyak jumlah data
    for i in range(n):
        swapped = False
        # Perulangan untuk membandingkan elemen yang bersebelahan
        for j in range(0, n - i - 1):
            # Jika nilai elemen ke-j LEBIH KECIL dari elemen ke-(j+1), tukar posisinya
            if arr[j]["nilai"] < arr[j + 1]["nilai"]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        
        # Jika tidak ada elemen yang ditukar, berarti data sudah urut
        if not swapped:
            break
            
    return arr

# ==========================================
# OUTPUT PROGRAM
# ==========================================
print("=== DATA MAHASISWA SEBELUM DIURUTKAN ===")
for mhs in data_mahasiswa:
    print(f"NIM: {mhs['nim']} | Nama: {mhs['nama']:<15} | Nilai: {mhs['nilai']}")

# Memanggil fungsi sorting
data_terurut = urutkan_nilai_tertinggi(data_mahasiswa)

print("\n=== DATA MAHASISWA SETELAH DIURUTKAN (NILAI TERTINGGI) ===")
for mhs in data_terurut:
    print(f"NIM: {mhs['nim']} | Nama: {mhs['nama']:<15} | Nilai: {mhs['nilai']}")
```

# Analisis Program Pengurutan Data Mahasiswa

Program ini dibuat untuk menyimpan dan mengurutkan data mahasiswa berdasarkan nilai tertinggi (descending) menggunakan struktur data **List of Dictionaries** dan algoritma **Bubble Sort**.

## 1. Struktur Data
Data mahasiswa disimpan dalam sebuah *List* bernama `data_mahasiswa`. Setiap elemen di dalam list tersebut berupa *Dictionary* yang menyimpan tiga informasi utama:
* `nama`: Menyimpan nama mahasiswa (String)
* `nim`: Menyimpan Nomor Induk Mahasiswa (String)
* `nilai`: Menyimpan nilai akhir mahasiswa (Integer/Float)

Terdapat 5 data mahasiswa yang dimasukkan sebagai nilai awal (inisialisasi) untuk memenuhi syarat minimal 5 mahasiswa.

## 2. Algoritma Sorting yang Digunakan
Sesuai dengan ketentuan bahwa tidak diperbolehkan menggunakan fungsi bawaan `sort()` atau `sorted()` dari Python, program ini menggunakan algoritma **Bubble Sort**.

**Cara Kerja Bubble Sort (Descending):**
1.  Program akan melakukan perulangan (iterasi) melewati seluruh elemen di dalam list.
2.  Pada setiap iterasi, program membandingkan elemen saat ini (indeks `j`) dengan elemen berikutnya (indeks `j + 1`).
3.  Karena kita ingin mengurutkan dari nilai **tertinggi ke terendah**, maka jika nilai elemen saat ini *lebih kecil* dari nilai elemen berikutnya (`arr[j]["nilai"] < arr[j+1]["nilai"]`), posisi kedua elemen tersebut akan **ditukar** (di-swap).
4.  Proses ini terus berulang hingga nilai terbesar "menggelembung" (berpindah) ke posisi paling awal list.
5.  Untuk optimasi, ditambahkan variabel `swapped`. Jika dalam satu iterasi penuh tidak ada pertukaran posisi sama sekali, itu berarti data sudah terurut dan perulangan bisa dihentikan lebih awal untuk menghemat waktu komputasi.
