### 1. Kode Program (Python)

Kode berikut menggunakan pendekatan *Backtracking* untuk mencari semua kemungkinan penempatan 8 ratu di papan catur berukuran 8 \times 8 sedemikian rupa sehingga tidak ada ratu yang saling serang.

```python
def solve_n_queens(N):
    # Inisialisasi papan catur N x N dengan angka 0
    board = [[0 for _ in range(N)] for _ in range(N)]
    solutions = []

    # Fungsi untuk mengecek apakah posisi ratu aman dari serangan
    def is_safe(board, row, col):
        # Cek baris horizontal ke arah kiri
        for i in range(col):
            if board[row][i] == 1:
                return False

        # Cek diagonal atas ke arah kiri
        for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
            if board[i][j] == 1:
                return False

        # Cek diagonal bawah ke arah kiri
        for i, j in zip(range(row, N, 1), range(col, -1, -1)):
            if board[i][j] == 1:
                return False

        return True

    # Fungsi rekursif utama untuk menempatkan ratu
    def solve_n_queens_util(board, col):
        # Basis kasus: Jika semua ratu berhasil ditempatkan
        if col >= N:
            solution = []
            for i in range(N):
                row_str = ""
                for j in range(N):
                    if board[i][j] == 1:
                        row_str += "Q " # Q merepresentasikan Ratu
                    else:
                        row_str += ". " # . merepresentasikan kotak kosong
                solution.append(row_str)
            solutions.append(solution)
            return True

        res = False
        # Coba tempatkan ratu di semua baris pada kolom (col) saat ini
        for i in range(N):
            if is_safe(board, i, col):
                # Tempatkan ratu
                board[i][col] = 1
                
                # Rekursi ke kolom berikutnya
                res = solve_n_queens_util(board, col + 1) or res
                
                # BACKTRACK: Jika penempatan tidak berujung pada solusi, hapus ratu
                board[i][col] = 0 

        return res

    # Mulai proses dari kolom 0
    solve_n_queens_util(board, 0)
    return solutions

# MENGUBAH NILAI N MENJADI 8
N = 8
solusi = solve_n_queens(N)

# Menampilkan Hasil
print(f"Total solusi yang ditemukan untuk {N}-Queens: {len(solusi)}\n")
print("Visualisasi Solusi Pertama:")
for baris in solusi[0]:
    print(baris)
```

### 2. Analisa Kodingan

Program ini dibagi menjadi beberapa komponen logika utama yang berpusat pada eksplorasi dan pemangkasan (*pruning*) kandidat solusi:

* **Inisialisasi Papan (`board`):** Papan direpresentasikan sebagai list 2 dimensi (matriks) berukuran $8 \times 8$. Nilai `0` menandakan petak kosong, sedangkan nilai `1` menandakan petak yang diisi oleh ratu (`Q`).
* **Fungsi Validasi (`is_safe`):** Karena program menempatkan ratu kolom demi kolom dari kiri ke kanan, kita hanya perlu mengecek potensi serangan dari sisi kiri ratu saat ini. Fungsi mengecek 3 arah: baris yang sama ke kiri, diagonal atas-kiri, dan diagonal bawah-kiri.
* **Fungsi *Backtracking* (`solve_n_queens_util`):**
    * **Penelusuran (Eksplorasi):** Fungsi mencoba menempatkan ratu di setiap baris `i` pada kolom `col`. Jika aman (`is_safe` mereturn `True`), ratu ditempatkan (`board[i][col] = 1`).
    * **Rekursi:** Program memanggil dirinya sendiri untuk kolom berikutnya (`col + 1`).
    * **Mundur (*Backtracking*):** Baris kode `board[i][col] = 0` adalah kunci algoritmanya. Jika rekursi menemui jalan buntu (tidak bisa menempatkan ratu di kolom berikutnya), program akan membatalkan langkah sebelumnya dengan mengubah nilai `1` kembali menjadi `0` dan mencoba baris lain.


### 3. Analisa Hasil AI (Perspektif Teknis & Komputasi)

Dari sudut pandang mesin dan optimasi algoritma, penyelesaian 8-Queens menggunakan kode di atas menghasilkan analisa matematis berikut:

* **Hasil Eksekusi:** Algoritma secara konsisten akan menemukan tepat **92 solusi unik** untuk papan catur berukuran $8 \times 8$.
* **Kompleksitas Waktu (*Time Complexity*):** Karena algoritma mengevaluasi setiap kemungkinan penempatan ratu di matriks, kompleksitas waktu *worst-case* mendekati $O(N!)$. Namun, berkat optimasi pengecekan `is_safe` yang memotong cabang eksplorasi yang tidak valid (*pruning*), jumlah iterasi aktual jauh lebih kecil dibandingkan *brute-force* murni sebesar $O(N^N)$.
* **Kompleksitas Ruang (*Space Complexity*):** Ruang memori yang dibutuhkan adalah $O(N^2)$ untuk menyimpan matriks papan catur. Selain itu, *call stack* pada proses rekursi memakan memori sebesar $O(N)$, sehingga efisiensi ruang program ini sangat optimal.
* **Kelebihan Algoritma:** Pendekatan *Depth-First Search* (DFS) yang digabungkan dengan backtracking terbukti stabil dan secara logis mencegah redundansi pengecekan yang berulang pada state papan yang sudah dipastikan salah.
