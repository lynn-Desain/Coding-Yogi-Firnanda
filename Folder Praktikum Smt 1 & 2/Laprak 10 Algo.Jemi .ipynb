```python
def solve_n_queens(N):
    # Inisialisasi papan catur N x N dengan angka 0
    board = [[0 for _ in range(N)] for _ in range(N)]
    solutions = []

    # Fungsi untuk mengecek apakah posisi ratu aman dari serangan
    def is_safe(board, row, col):
        # Cek baris horizontal ke arah kiri
        for i in range(col):
            if board[row][i] == 1:
                return False

        # Cek diagonal atas ke arah kiri
        for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
            if board[i][j] == 1:
                return False

        # Cek diagonal bawah ke arah kiri
        for i, j in zip(range(row, N, 1), range(col, -1, -1)):
            if board[i][j] == 1:
                return False

        return True

    # Fungsi rekursif utama untuk menempatkan ratu
    def solve_n_queens_util(board, col):
        # Basis kasus: Jika semua ratu berhasil ditempatkan
        if col >= N:
            solution = []
            for i in range(N):
                row_str = ""
                for j in range(N):
                    if board[i][j] == 1:
                        row_str += "Q " # Q merepresentasikan Ratu
                    else:
                        row_str += ". " # . merepresentasikan kotak kosong
                solution.append(row_str)
            solutions.append(solution)
            return True

        res = False
        # Coba tempatkan ratu di semua baris pada kolom (col) saat ini
        for i in range(N):
            if is_safe(board, i, col):
                # Tempatkan ratu
                board[i][col] = 1
                
                # Rekursi ke kolom berikutnya
                res = solve_n_queens_util(board, col + 1) or res
                
                # BACKTRACK: Jika penempatan tidak berujung pada solusi, hapus ratu
                board[i][col] = 0 

        return res

    # Mulai proses dari kolom 0
    solve_n_queens_util(board, 0)
    return solutions

# MENGUBAH NILAI N MENJADI 8
N = 8
solusi = solve_n_queens(N)

# Menampilkan Hasil
print(f"Total solusi yang ditemukan untuk {N}-Queens: {len(solusi)}\n")
print("Visualisasi Solusi Pertama:")
for baris in solusi[0]:
    print(baris)
```

### 1. Analisis AI (Pendekatan Sistem, Kompleksitas, & Paradigma)
Bagian ini menganalisis kode dari sudut pandang ilmu komputer teoretis dan efisiensi mesin.

* **Paradigma Algoritma:** Kode ini menggunakan algoritma *Backtracking* yang berjalan di atas struktur *Depth-First Search* (DFS). Algoritma ini mencoba menelusuri setiap kemungkinan penempatan ratu secara mendalam ke setiap kolom, dan akan mundur (*backtrack*) jika menemukan jalan buntu (kondisi di mana ratu saling serang).
* **Kompleksitas Waktu (Time Complexity):** Di atas kertas, kompleksitas terburuk dari N-Queens adalah $O(N!)$. Untuk setiap kolom, fungsi mencoba $N$ baris, lalu $N-1$ baris di kolom berikutnya, dan seterusnya. Untuk $N=8$, mesin masih dapat menyelesaikannya dalam hitungan milidetik karena jumlah permutasinya dapat ditangani dengan cepat, tetapi akan sangat berat jika nilai $N$ diperbesar (misalnya $N=20$).
* **Kompleksitas Ruang (Space Complexity):** $O(N^2)$ untuk menyimpan matriks `board` (papan catur ukuran $N \times N$), ditambah $O(N)$ untuk *call stack* memori akibat rekursi sedalam $N$. Secara keseluruhan, kompleksitas ruangnya didominasi oleh $O(N^2)$.
* **Evaluasi Keamanan (is_safe):** Logika pengecekan di dalam fungsi `is_safe` sudah sangat teroptimasi. Mesin tidak perlu mengecek seluruh papan. Karena algoritma mengisi papan dari kolom kiri ke kanan, mesin hanya perlu mengecek ancaman dari arah kiri (horizontal kiri, diagonal atas-kiri, dan diagonal bawah-kiri). Sisi kanan diabaikan karena pasti belum ada ratu yang ditempatkan di sana.

### 2. Analisis Sendiri (Pendekatan Manual & Alur Logika)
Bagian ini adalah pembedahan alur logika seolah-olah kita sedang menelusuri (*tracing*) jalannya kode secara manual baris demi baris.

* **Proses Inisialisasi:** Pertama, program menyiapkan `board` kosong berupa matriks 2D berukuran 8x8 yang diisi dengan angka 0. Variabel `solutions` disiapkan dalam bentuk *list* kosong untuk menampung semua papan yang berhasil memenuhi syarat.
* **Langkah Awal (Kolom 0):**
    Fungsi `solve_n_queens_util(board, 0)` dipanggil. Program mencoba menempatkan Ratu pertama di baris 0, kolom 0. Karena `is_safe` mengembalikan `True` (papan masih kosong), Ratu ditempatkan (`board[0][0] = 1`). Lalu, program memanggil dirinya sendiri (*rekursi*) untuk maju ke kolom 1.
* **Mekanisme "Trial and Error":**
    Di kolom 1, program mencoba baris 0. Pengecekan `is_safe` akan gagal karena ada Ratu di horizontal kiri. Program lanjut mencoba baris 1. Pengecekan gagal lagi karena ada Ratu di diagonal atas-kiri. Program baru berhasil menaruh Ratu di baris 2 atau 3. Proses penempatan dan pengecekan ini terus berlanjut.
* **Fase Backtracking (Mundur Tepat Waktu):**
    Ini adalah inti dari kode tersebut. Perhatikan baris `board[i][col] = 0`. Jika penempatan Ratu di suatu sel ternyata berujung pada jalan buntu di kolom-kolom berikutnya (fungsi rekursi mengembalikan `False`), program tidak *error* atau berhenti. Alih-alih, ia mencabut kembali Ratu tersebut (mengubahnya dari 1 menjadi 0), lalu mencoba baris berikutnya.
* **Panen Solusi:**
    Ketika `col >= N` (mencapai kolom ke-8, yang berarti indeks 0-7 sudah terisi semua), itu artinya 8 Ratu sudah berhasil diletakkan tanpa saling serang. Di titik ini, program melakukan iterasi untuk mengubah angka 1 menjadi huruf "Q" dan angka 0 menjadi titik ".", lalu menggabungkannya menjadi *string* visual untuk disimpan ke dalam list `solutions`. Untuk $N=8$, kode ini akan mengumpulkan total 92 solusi berbeda.
